```python
import numpy as np
import pandas as pd

np.random.seed(42)

ids = np.arange(1, 11)
ages = np.random.randint(18, 60, 10)
salaries = np.random.randint(30000, 90000, 10)
departments = np.array(["HR", "IT", "Finance", "IT", "HR", "Sales", "Finance", "IT", "Sales", "HR"])

DF = pd.DataFrame({
"id": ids,
"age": ages,
"salary": salaries,
"dept": departments
})

DF.to_csv("employees.csv", index=False)
print("Sample Data Created and Saved as employees.csv")
```

```python

df = pd.read_csv("employees.csv")

ages = np.array(df['age'])
salaries = np.array(df['salary'])

print(ages)
print(salaries)

print(ages.dtype)
print(ages.ndim)
print(ages.shape)
print(ages.size)


print(salaries.dtype)
print(salaries.ndim)
print(salaries.shape)
print(salaries.size)

```

```python
#Using the salaries array, found and printed the highest and lowest salary values.
#Also, calculated and printed the average salary and age using NumPy functions.
highest_salary = np.max(salaries)
lowest_salary = np.min(salaries)
average_salary = np.mean(salaries)
print("Highest salary: ",highest_salary)
print("Lowest salary: ",lowest_salary)
average_age = np.mean(ages)
print("Average salary: ",average_salary)
print("Average age: ",average_age)
```

```python
#From the ages array, filtered and printed all ages greater than 30 using a boolean condition.
#Then counted and printed how many employees are older than 30.
ages_over_30 = ages[ages > 30]
count_over_30 = np.sum(ages > 30)
print("Ages greater than 30:", ages_over_30)
print("Number of employees older than 30:", count_over_30)
```

```python
#Created a new NumPy array that increases every employee’s age by 5 years. Printed the new updated ages array.
updated_ages = ages + 5
print(updated_ages)
```

```python
#Using NumPy, calculated the total salary expense and the difference between the maximum and minimum salary. Printed both results
total_salary = np.sum(salaries)
salary_diff = np.max(salaries) - np.min(salaries)
print("Total Salary Expense:", total_salary)
print("Difference:", salary_diff)
```

Q6. Load the employees.csv file into a Pandas DataFrame. Display the first 5 rows, and check basic info using info() and summary statistics using describe().


```python
#Loaded the employees.csv file into a Pandas DataFrame. Displayed the first 5 rows, and checked basic info and summary statistics
import pandas as pd
df = pd.read_csv('employees.csv')
print(df.head())
print(df.info())
print(df.describe())
```

```python
#Displayed  the id, age, and salary columns. Then showed the last 3 rows
selected_columns = df[['id', 'age', 'salary']]
print(selected_columns)
print(selected_columns.tail(3))
```

```python
#Filtered the DataFrame to show only employees who work in the IT department
it_employees = df[df['dept'] == 'IT']
print(it_employees)
print("total no. of IT employees: ",len(it_employees))
```

Q9. Sort the DataFrame by salary in descending order and display the top 3 highest-paid employees along with their department and age.


```python
# Sorted the DataFrame by salary in descending order and displayed the top 3 highest-paid employees along with their department and age
sorted_df = df.sort_values(by='salary', ascending=False)
top_3 = sorted_df[['id', 'dept', 'age', 'salary']].head(3)
print(top_3)
```

Q10. Replace all salary values greater than 80000 with 80000 using loc. Then calculate and print the new average salary of all employees after replacement.


```python
#Replaced all salary values greater than 80000 with 80000
#calculated and printed the new average salary of all employees after replacement
df.loc[df['salary'] > 80000, 'salary'] = 80000
average_salary = df['salary'].mean()
print("New average salary", average_salary)
```
